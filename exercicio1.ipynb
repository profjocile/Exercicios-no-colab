# Exercícios de programação

1. Faça um algoritmo para ler dois números e imprimir a soma dos números lidos.

### Algoritmo da soma

1. Leia o primeiro número e guarde numa variável chamada numero1
2. Leia o segundo número e guarde em outra variável chamada numero2
3. Faça a soma das duas variáveis e guarde o resultado em outra variável chamada soma
4. mostre o resultado

```python
numero1 = int(input('Informe o 1º número'))
numero2 = int(input('Informe o 2º número'))
print(type(numero1))
soma = numero1 + numero2
print(f'O resultado é: {soma}')
```

### Algoritmo da média

2. Faça um algoritmo para ler três números e imprimir a soma, média e produto dos números lidos.

Algoritmo:

1. Leia o primeiro número e salve na variável num1
2. Leia o segundo número e salve na variável num2
3. Leia o terceiro número e salve na variável num3
4. Calcule a soma dos 3 números e salve na variável soma
5. Calcule a média dos 3 números e salve na variável media
6. Calcule o produto dos 3 números e salve na variável produto
7. Mostre o resultado da soma
8. Mostre o resultado da média
9. Mostre o resultado do produto

```python
num1 = int(input('Informe o 1º número: '))
num2 = int(input('Informe o 2º número: '))
num3 = int(input('Informe o 3º número: '))

soma = num1 + num2 + num3
media = soma / 3
produto = num1 * num2 * num3

print(f'A soma dos números é: {soma}')
print(f'A média dos números é: {media:.2f}')
print(f'O produto dos números é: {produto}')
```

### Algoritmo do maior e menor

3. Faça um algoritmo para ler dois números e imprimir o maior, o menor ou então dizer se são iguais.

Algoritmo:

1. Leia o primeiro número e salve na variável num1
2. Leia o segundo número e salve na variável num2
3. Descobrir o maior
4. Mostrar o maior
5. Descobrir o menor
6. Mostrar o menor
7. Descobir se são iguais
8. Mostrar se são iguais

```python
num1 = int(input('Informe o 1º número: '))
num2 = int(input('Informe o 2º número: '))
if num1 > num2:
    print(f'O maior número é: {num1}')
    print(f'O menor número é: {num2}')
elif num2 > num1:
    print(f'O maior número é: {num2}')
    print(f'O menor número é: {num1}')
else:
    print('Os números são iguais')
```
