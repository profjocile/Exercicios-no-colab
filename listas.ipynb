### Questão 1. Crie um programa que receba uma lista de números e retorne o maior valor dessa lista.

Algoritmo:

1. Criar uma lista com vários números

```python
lista_de_numeros = [10, 5, 9, 11, 10, 3]
```

2. Percorrer a lista em um loop para salvar o maior valor em uma variável


```python
maior_valor_da_lista = 0
for numero in lista_de_numeros:
  if numero > maior_valor_da_lista:
    maior_valor_da_lista = numero
print(f'''
        O maior valor da lista:
        {lista_de_numeros} é {maior_valor_da_lista}
      ''')
```

### Questão 2. Crie um programa que receba uma lista de números e retorne a soma de todos os valores.

Algoritmo:
1. Cria a lista de números
2. Fazer um loop para receber os números e colocar na lista
3. Somar os números em uma variável soma
4. Mostrar a soma dos números

```python
lista_de_numeros = []
soma = 0
repetir = True
while repetir:
  numero = int(input('Entre com um número: '))
  lista_de_numeros.append(numero)
  soma = soma + numero
  continuar = input('Deseja continuar? [s/n]')
  if continuar == 'n':
    repetir = False
print(lista_de_numeros)
print(f'A soma é {soma}')
```

### Questão 3. Crie um programa que receba duas listas e retorne uma terceira lista com a interseção dos elementos das duas listas.

Algoritmo:

1. Receber uma lista
2. Receber outra lista
3. Percorrer a primeira e lista e comparar os elementos com os elementos da outra lista e salvar os elementos repetidos em uma terceira lista
4. Mostrar a terceira lista

```python
from ast import Break
primeira_lista = []
segunda_lista = []
terceira_lista = []
repetir = True
while True:
  numero = int(input('Entre com um número para a primeira lista: '))
  primeira_lista.append(numero)
  continuar = input('Deseja continuar? [s/n]')
  if continuar == 'n':
    break
while repetir:
  numero = int(input('Entre com um número para a segunda lista: '))
  segunda_lista.append(numero)
  continuar = input('Deseja continuar? [s/n]')
  if continuar == 'n':
    repetir = False
for numero in primeira_lista:
  if numero in segunda_lista:
    terceira_lista.append(numero)
print(f'A primeira lista é {primeira_lista}')
print(f'A segunda lista é {segunda_lista}')
print(f'A interção é {terceira_lista}')
```
